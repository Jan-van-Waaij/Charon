```python
import msprime
```

```python
demography = msprime.Demography()
```

```python
demography.add_population(name="A", initial_size=10_000, growth_rate=0.0)
```

```python
demography.add_population(name="B", initial_size=5_000, growth_rate=0.0)
```

```python
demography.add_population(name="C", initial_size=1_000, growth_rate=0.0)
```

```python
demography.add_population_split(time=1000, derived=["A", "B"], ancestral="C")
```

```python
ts = msprime.sim_ancestry(samples={"A":5_000, "B":5_000}, demography=demography, random_seed=12, recombination_rate=1e-8, sequence_length=10e6)
```

```python
ts
```

```python
mts = msprime.sim_mutations(ts, rate=2e-8)
# mutation_rate=2e-8 recombination_rate=1e-8
```
