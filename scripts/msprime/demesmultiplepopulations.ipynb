```python
import msprime
```

```python
import demes
```

```python
import os
os.getcwd()
```

```python
graph = demes.load("../../data/demesfiles/oneancientfoursplits.yaml")
```

```python
demography = msprime.Demography.from_demes(graph)
```

```python
ts = msprime.sim_ancestry(samples={"common_ancestor":5_000, "anchor1":5_000, "anchor2":5_000, "anchor3":5_000, "anchor4":5_000}, demography=demography, random_seed=12, recombination_rate=1e-8, sequence_length=10e6)
```

```python
mts = msprime.sim_mutations(ts, rate=2e-8)
```

```python
with open("test.vcf", "w") as vcf_file:
        mts.write_vcf(vcf_file)
```

```python
mts.dump("test.tree")
```
